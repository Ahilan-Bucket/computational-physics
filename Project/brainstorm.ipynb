## Core: Numerical Flood Model (Week 1 – 1.5)

### 1. Terrain Preprocessing
- Load Cooum STL mesh using `numpy-stl`
- Ray-cast the 3D mesh onto a regular 2D grid to extract elevation $z(x, y)$
- Visualize with contour plot and 3D surface plot

### 2. PDE Solver — Shallow-Water Diffusion

**Governing equation:**

$$\frac{\partial h}{\partial t} = \nabla \cdot \left[ D(h) \, \nabla(h + z) \right] + R(t)$$

where:
- $h(x, y, t)$ = water depth
- $z(x, y)$ = terrain elevation (from STL)
- $D(h)$ = diffusion coefficient (function of depth, e.g. Manning's formula)
- $R(t)$ = rainfall forcing (time-varying source term)

**Discretization:**
- **Spatial:** Central-difference FDM on uniform grid (connects to **Guide 3** — Taylor expansion stencils)
- **Temporal:** Forward Euler with CFL stability condition (connects to **Guide 4** — ODE integration)
- **Boundary conditions:** Open outflow at river mouth, no-flow at basin edges

### 3. Rainfall Forcing
- Load Chennai daily rainfall CSV (1991–2023)
- Extract **December 1–3, 2015** peak event (291.5 mm on Dec 2)
- Apply as spatially uniform, time-varying source term $R(t)$

### 4. Validation & Diagnostics
- **Mass conservation:** Verify $\int \!\!\int h \, dA = \int R \, dt$ using trapezoidal integration
- **Flat-terrain sanity test:** Should produce symmetric radial spreading (analytical check)

### 5. Visualization
- Flood inundation maps at multiple timesteps (`contourf` over terrain)
- Animated water depth evolution (`matplotlib.animation`)
- Mass conservation diagnostic plot (total water volume vs. time)

### 6. Numerical Analysis
- **Grid convergence study:** Run at 32x32, 64x64, 128x128
- Plot $L_2$ error norm vs. $\Delta x$ on log-log scale, extract order of convergence
- **CFL stability demonstration:** Show numerical blow-up when CFL is violated, stable solution when satisfied
- **Optional:** Compare Forward Euler vs. RK2 time-stepping accuracy

## Reach A: Physics-Informed Neural Network — PINN (~3 days)

### What Is a PINN?
A neural network trained to solve the PDE **without a grid**. Instead of supervised data, the **loss function encodes the physics** — the PDE residual, boundary conditions, and initial conditions are evaluated at random collocation points using automatic differentiation.

The PINN *is* the digital twin: a learned, differentiable proxy of the physical system.

### Architecture
- **Inputs:** $(x, y, t)$ — spatial coordinates + time
- **Network:** 4 hidden layers, 64 neurons each, $\tanh$ activation
- **Output:** $\hat{h}(x, y, t)$ — predicted water depth
- **Framework:** PyTorch (from scratch)

### Loss Function

$$\mathcal{L} = \lambda_{\text{PDE}} \mathcal{L}_{\text{PDE}} + \lambda_{\text{BC}} \mathcal{L}_{\text{BC}} + \lambda_{\text{IC}} \mathcal{L}_{\text{IC}} + \lambda_{\text{data}} \mathcal{L}_{\text{data}}$$

| Term | Description |
|---|---|
| $\mathcal{L}_{\text{PDE}}$ | PDE residual at ~2000–5000 random collocation points (derivatives via `autograd`) |
| $\mathcal{L}_{\text{BC}}$ | Boundary condition violation |
| $\mathcal{L}_{\text{IC}}$ | Initial condition violation (dry terrain at $t=0$) |
| $\mathcal{L}_{\text{data}}$ | Optional: a few points from the FDM solution for guidance |

**Training:** Adam optimizer, ~5000–10000 epochs

### The Comparison (The Impressive Part)

| Metric | FDM (Core) | PINN (Reach) |
|---|---|---|
| Flood maps | Side-by-side at same timesteps | |
| Speed | Wall-clock solve time | Training time vs. inference time |
| Mass conservation | Trapezoidal check | Does the PINN conserve mass? |
| Accuracy | Reference solution | $L_2$ error vs. FDM |

**Key discussion:** When does the PINN win? Tradeoffs between training cost, inference speed, generalization to unseen rainfall scenarios, and physics fidelity.

## Reach B: Financial Risk Estimation (~1 day)

### Depth-Damage Model
Use standard **FEMA/USACE depth-damage functions** from literature:

$$\text{damage}(\$) = f\!\left(h_{\text{flood}},\, \text{building type}\right)$$

Typical form: piecewise-linear or exponential curve, calibrated to South Asian construction costs.

### Risk Map
- Define a simple **land-use grid** over the Cooum basin (residential, commercial, infrastructure)
- For each grid cell: $\text{flood depth} \times \text{depth-damage curve} \rightarrow \text{estimated \$ loss}$
- Produce a **spatial risk map** showing damage per cell
- Compute **total basin damage** estimate for the 2015 event
